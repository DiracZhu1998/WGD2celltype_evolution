```r
suppressMessages({
    require(Seurat)
    require(dplyr)
})


get_coding <- function(atlas, coding_genes){
    subset <- atlas@assays$RNA@counts[coding_genes, ]
    subset <- CreateSeuratObject(subset, meta.data = atlas@meta.data)
    subset <- NormalizeData(subset, normalization.method = "LogNormalize", scale.factor = 10000)
    return(subset)
}

getUMAP <- function(object){
    object <- FindVariableFeatures(object, selection.method = "vst", nfeatures = 2000)
    object <- ScaleData(object)
    object <- RunPCA(object, npcs = 20)
    object <- FindNeighbors(object, dims = 1:10)
    object <- RunUMAP(object, dims=1:10)
    return(object)
}
```

```r
# for mouse
info <- read.delim('../2.ohnolog_significance/2.bg/mouse.info', header = T)
info <- info %>% filter(Gene.type == 'protein_coding') 
info[info$Gene.name == '', 'Gene.name'] <- info[info$Gene.name == '', 'Gene.stable.ID']

load('GSM4873765_mouse_data.RData')
mouse <- RL
DefaultAssay(mouse) <- 'RNA'
mouse@meta.data$cell_type <- sub("\\d+$", "", mouse@meta.data$final.clusters2)
mouse@meta.data$classes <- sub(".*\\.()", "", mouse@meta.data$cell_type)
mouse <- UpdateSeuratObject(mouse)
coding_genes = rownames(mouse)[rownames(mouse) %in% info$Gene.name]
mouse <- get_coding(mouse, coding_genes)
mouse <- getUMAP(mouse)
saveRDS(mouse, file = 'GSM4873765_mouse_exc.coding_genes.rds')

mouse <- VZ
DefaultAssay(mouse) <- 'RNA'
mouse@meta.data$cell_type <- mouse@meta.data$final.clusters2
mouse@meta.data$classes <- 'Inh'
mouse <- UpdateSeuratObject(mouse)
coding_genes = rownames(mouse)[rownames(mouse) %in% info$Gene.name]
mouse <- get_coding(mouse, coding_genes)
mouse <- getUMAP(mouse)
saveRDS(mouse, file = 'GSM4873765_mouse_inh.coding_genes.rds')
```
