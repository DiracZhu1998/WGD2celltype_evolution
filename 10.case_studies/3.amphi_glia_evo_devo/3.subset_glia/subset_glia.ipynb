```r
# subset glial cells for N4, T1, adult brain, and adult neural tube
suppressMessages({
    require(Seurat)
})
```

```r
N4 <- readRDS('/mnt/data01/yuanzhen/01.Vertebrate_cell_evo/01.data/02.atlas/amphioxus/N4_stage_subset.rds')
T1 <- readRDS('/mnt/data01/yuanzhen/01.Vertebrate_cell_evo/01.data/02.atlas/amphioxus/T1_stage_subset.rds')
adult_brain <- readRDS('/mnt/data01/yuanzhen/03.sensory_neurosecretory_evo/01.data/03.atlas/Bflo/brain_integrated/4000_40_30/RNA_harmony_integrated.RDS')
adult_nt <- readRDS('/mnt/data01/yuanzhen/03.sensory_neurosecretory_evo/01.data/03.atlas/Bflo/neural_tube_adult_BD/BD_nt.rds')
```

```r
head(adult_brain@meta.data)
```

```r
Idents(N4) <- 'celltype'
N4_glia <- subset(N4, idents = c('N4:RhSp OligB+ cells_60', 'N4:Posterior RhSp cells_4', 'N4:RhSp Dlx+ cells_41', 'N4:RhSp FoxB2+ cells_18'))
Idents(T1) <- 'celltype'
T1_glia <- subset(T1, idents = c('T1:Posterior RhSp cells_35', 'T1:Posterior RhSp cells_19', 'T1:Posterior floor plate tip cells_66',
                                'T1:Posterior RhSp tip Dlx+ cells_56', 'T1:Anterior floor plate cells_32', 'T1:Posterior RhSp cells_46',
                                'T1:Posterior RhSp cells_65', 'T1:RhSp Dlx+ cells_86', 'T1:DiMes cells_76'))
```

```r
adult_brain@meta.data$stage <- 'adultbrain'
adult_brain@meta.data$celltype <- paste('adultbrain', adult_brain@meta.data$`RNA_snn_res.1`, sep = '_')
adult_brain_glia <- subset(adult_brain, subset = RNA_snn_res.1 %in% c(4,8,12,5,20,6,3))
Idents(adult_brain_glia) <- 'celltype'
```

```r
adult_nt@meta.data$stage <- 'adultnt'
adult_nt@meta.data$celltype <- paste('adultnt', adult_nt@meta.data$`RNA_snn_res.1`, sep = '_')
adult_nt_glia <- subset(adult_nt, idents = c(0,4,3,8,5,7,6,9,2,15))
Idents(adult_nt_glia) <- 'celltype'
```

```r
saveRDS(N4_glia, 'N4_glia.rds')
saveRDS(T1_glia, 'T1_glia.rds')
saveRDS(adult_brain_glia, 'adult_brain_glia.rds')
saveRDS(adult_nt_glia, 'adult_nt_glia.rds')
```
